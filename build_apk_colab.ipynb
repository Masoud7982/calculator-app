## ۱) آپلود فایل `calculator.zip`
دکمه‌ی Choose Files را بزنید و فایل `calculator.zip` را از پوشه‌ی پروژه انتخاب کنید.

```python
from google.colab import files
uploaded = files.upload()  # فایل calculator.zip را انتخاب کنید
```

```python
import zipfile, os

zip_name = next(iter(uploaded))
os.makedirs("project", exist_ok=True)
with zipfile.ZipFile(zip_name) as zf:
    zf.extractall("project")
os.chdir("project")
!ls -la
```

## ۳) دانلود فایل APK

```python
import glob
from google.colab import files

apks = glob.glob("bin/*.apk")
assert apks, "هیچ APK ساخته نشد — لاگ سلول قبل را بررسی کنید"
print("فایل ساخته‌شده:", apks[0])
files.download(apks[0])
```
